```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from surprise import Dataset
from surprise import Reader
from surprise import SVD
from surprise.model_selection import train_test_split
from surprise import accuracy

import warnings
warnings.filterwarnings("ignore")
```

```python
df = pd.read_csv("Datafiniti_Amazon_Consumer_Reviews_of_Amazon_Products.csv")
```

```python
recommend_df = df[['reviews.username', 'name', 'reviews.rating']]
recommend_df.head()
```

```python
recommend_df.isnull().sum()
```

```python
recommend_df = recommend_df.dropna()
recommend_df.shape
```

```python
recommend_df.columns = ['user', 'product', 'rating']
recommend_df.head()
```

```python
print("Total Users:", recommend_df['user'].nunique())
print("Total Products:", recommend_df['product'].nunique())
```

```python
reader = Reader(rating_scale=(1,5))

data = Dataset.load_from_df(
    recommend_df[['user','product','rating']],
    reader
)
```

```python
trainset, testset = train_test_split(data, test_size=0.2, random_state=42)
```

```python
model = SVD()

model.fit(trainset)
```

```python
predictions = model.test(testset)
```

```python
accuracy.rmse(predictions)
```

```python
def recommend_products(user_name, n=5):
    products = recommend_df['product'].unique()

    predictions = []

    for product in products:
        pred = model.predict(user_name, product)
        predictions.append((product, pred.est))

    recommendations = sorted(predictions, key=lambda x: x[1], reverse=True)

    return recommendations[:n]
```

```python
recommend_products("Illyyue")
```

```python
from surprise.model_selection import cross_validate

cross_validate(model, data, measures=['RMSE', 'MAE'], cv=5, verbose=True)
```
